1. push

3. peep

User will be given choice to perform action from above three.
Script should be kept running until user chooses to exit

```python
stack=list()
current_count=1

def push():
   global current_count
   if current_count<=count: 
    num=int(input(f"Enter Stack Element :  {current_count}/{count}"))
    current_count+=1
    stack.append(num)
   else:
     print("Stack Is Overflow : ")

def pop():
    global current_count
    if current_count==1:
        print("Stack Is Underflow (Empty) : ")
    else:
        stack.pop()
        current_count-=1
        print("POP Element : ")

def show():
    print(stack)

def peek():
    if current_count == 1:
        print("Stack Is Underflow (Empty) : ")
    else:
        print(stack[len(stack)-1])

        

count=int(input("Enter Stack Size : "))

while 1:
    print("1 For PUSH() \n 2 For POP() \n 3 For SHOW() \n 4 For PEEK() \n 5 For Exit")
    user=int(input("Enter Option : "))

    if user == 1:
     push()
    elif user == 2:
     pop()
    elif user == 3:
     show()
    elif user == 4:
      peek()
    elif user == 5:
        break
    else:
        print("Enter Valid Option : ")
```

## 2) Queue 

```python
Implement Queue
You script should contain three functions
1. enqueue
2. dequeue
3. peek
User will be given choice to perform action from above three.
Script should be kept running until user chooses to exit.
```

```python
list1=list()
size=int(input("Enter Queue Size : "))
current_size=1

def enqueue():
    global current_size
    global size
    if  current_size<=size:
        num=int(input(f"Enter Element : {current_size}/{size}"))
        list1.append(num)
        current_size+=1
    else:
        print("Queue Overflow : ")


def dequeue():
    global current_size
    global size
    if current_size==1:
        print("Queue in Underflow : ")
    else:
        ele=list1.pop(0)
        current_size-=1
        print(f"Dequeue Element : {ele}")
        

def show():
    print(list1)


def peek():
    if current_size==1:
        print("Queue Is Underflow")
    else:
        print(f"Peek Element :{list1[0]}")



while 1:
    print("Choose Option \n 1 For Enqueue \n 2 For Dequeue \n 3 For Show \n 4 For Peek \n 5 For Exit")
    num=int(input("Enter Number : "))

    if num == 1:
        enqueue()
    elif num==2:
        dequeue()
    elif num==3:
        show()
    elif num==4:
        peek()
    elif num==5:
        break
    
```

## 3) Circular Queue

```python
Implement Circular Queue
You script should contain four functions
1. Front: Get the front item from queue.
2. Rear: Get the last item from queue.
3. enqueue(value)
4. dequeue()
User will be given choice to perform action from above four.
Script should be kept running until user chooses to exit.
```

```python
list1=list()

size=int(input("Enter Size Of Circular Queue : "))
current_size=1


def FrontEnqueue():
    global current_size
    if current_size<=size:
        num=int(input(f"Enter Element (Front-Add) : {current_size}/{size}"))
        list1.insert(0,num)
        current_size+=1
    else:
        print("Circular Queue Is Overflow : ")


def RearEnqueue():
    global current_size
    if current_size<=size:
        num=int(input(f"Enter Element (Rear-Add) : {current_size}/{size}"))
        list1.append(num)
        current_size+=1
    else:
        print("Circular Queue Is Overflow : ")


def FrontDequeue():
    global current_size
    if current_size==1:
        print("Circular Queue Is Underflow : ")
    else:
        n=list1.pop(0)
        print(f"Front Dequeue : {n}")
        current_size-=1


def RearDequeue():
    global current_size
    if current_size==1:
        print("Circular Queue Is Underflow : ")
    else:
        n=list1.pop()
        print(f"Rear Dequeu : {n}")
        current_size-=1


def show():
    print(list1)

while 1:
    print("Choose Number \n 1 For FrontEnqueue \n 2 For RearEnqueue \n 3 For FrontDequeue \n 4 For RearDequeue \n 5 For Show \n 6 For Exit")
    user=int(input("Enter Option : "))

    if user==1:
        FrontEnqueue()
    elif user==2:
        RearEnqueue()
    elif user==3:
        FrontDequeue()
    elif user==4:
        RearDequeue()
    elif user==5:
        show()
    elif user==6:
        break 
    else:
        print("Enter Valid Option : ")
        
        
```

## 4) Selection Sort

```python
Selection sort
Use a static list
Use selection sort to sort this list
```

```python
def selection_sort(items):
    n = len(items)
    
    for i in range(n - 1):
        min_idx = i
        for j in range(i + 1, n):
            if items[j] < items[min_idx]:
                min_idx = j
                
        if min_idx != i:
            items[i], items[min_idx] = items[min_idx], items[i]
            
    return items

my_list = [64, 34, 25, 12, 22, 11, 90]
sorted_list = selection_sort(my_list)
print("Sorted List:", sorted_list)
```

## 5) Bubble Sort

```python
Bubble sort
Use a static list
Use bubble sort to sort this list
```

```python
def bubble_sort(arr):
    n = len(arr)
    
    for i in range(n):
        swapped = False
        

        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
                
        if not swapped:
            break

numbers = [64, 34, 25, 12, 22, 11, 90]
print("Original list:", numbers)

bubble_sort(numbers)
print("Sorted list:  ", numbers)
```

## 6) Insertion Sort

```python
Insertion sort
Use a static list
Use insertion sort to sort this list
```

```python
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i] 
        j = i - 1    
      
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
            
       
        arr[j + 1] = key


data = [12, 11, 13, 5, -1]
insertion_sort(data)
print("Sorted array:", data)
```

## 7) Binary Search

```python
Binary search
Use a static list
Use Binary search to find element asked by user
Use one of the sorting algorithm that you have developed to sort list here
```

```python
def binary_search(arr, target):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2
        

        if arr[mid] == target:
            return mid
       
        elif arr[mid] < target:
            low = mid + 1
      
        else:
            high = mid - 1
            
   
    return -1


mylist = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
print(binary_search(mylist, 23))  
```

## 9) Diterminent

## 12) ATM Machine

```python
Create an ATM machine
- User should be able get money out of it
- User should be able to deposite money into it
- You must ask card number and pin before any transaction
- User can get overall information about his bank account
- There should be a limit in one time transaction and total transaction in a day (Consider day as from
starting of script to exit)
- Limit on number of transaction
- Let user choose the bank of ATM
- Every ATM will have some initial balance
- Every user has it's own bank
- If user takes money from an ATM other than his bank take 5% cut in withdrawl amount
Nice to have features:
- Insert, update and delete users (Generate card number and PIN randomly)
- Insert, update and delete ATMs
- Insert, update and delete banks
- Insert money into ATM (No database is needed for this it can be managed using dicts)
Hint: You can manage all this with data dictionary data type.
```

```python
# # Create an ATM machine
# # - User should be able get money out of it
# # - User should be able to deposite money into it
# # - You must ask card number and pin before any transaction
# # - User can get overall information about his bank account
# # - There should be a limit in one time transaction and total transaction in a day (Consider day as from
# # starting of script to exit)
# # - Limit on number of transaction
# # - Let user choose the bank of ATM
# # - Every ATM will have some initial balance
# # - Every user has it's own bank
# # - If user takes money from an ATM other than his bank take 5% cut in withdrawl amount
# # Nice to have features:
# # - Insert, update and delete users (Generate card number and PIN randomly)
# # - Insert, update and delete ATMs
# # - Insert, update and delete banks
# # - Insert money into ATM (No database is needed for this it can be managed using dicts)
# # Hint: You can manage all this with data dictionary data type.

class Banks:

    def __init__(self, BankName, ATMBalance,
                 NumberOfTransactionLimit, OneDayLimit,
                 PerTransactionLimit):

        self.BankName = BankName
        self.ATMBalance = ATMBalance
        self.NumberOfTransactionLimit = NumberOfTransactionLimit
        self.OneDayLimit = OneDayLimit
        self.PerTransactionLimit = PerTransactionLimit

    def Bank_Detail(self):

        print("----------------------------------")
        print(f"Bank Name : {self.BankName}")
        print(f"ATM Balance : {self.ATMBalance}")
        print(f"Number Of Transaction Limit : {self.NumberOfTransactionLimit}")
        print(f"One Day Limit : {self.OneDayLimit}")
        print(f"Per Transaction Limit : {self.PerTransactionLimit}")
        print("----------------------------------")


class User:

    def __init__(self, UserName, BankName, Balance,
                 TransactionCount, CardNumber, PIN, WithDrawMoney):

        self.UserName = UserName
        self.BankName = BankName
        self.Balance = Balance
        self.TransactionCount = TransactionCount
        self.CardNumber = CardNumber
        self.PIN = PIN
        self.WithDrawMoney = WithDrawMoney

    def User_Detail(self):

        print("----------------------------------")
        print(f"User Name : {self.UserName}")
        print(f"Bank Name : {self.BankName}")
        print(f"Balance : {self.Balance}")
        print(f"Transaction Count : {self.TransactionCount}")
        print(f"Card Number : {self.CardNumber}")
        print(f"PIN Number : {self.PIN}")
        print(f"Total Withdraw Money : {self.WithDrawMoney}")
        print("----------------------------------")

    # Withdraw Money

    def WithDraw(self, CurrentWithDrawMoney,
                 Current_Bank_OneDayLimit,
                 Current_Bank_ATMBalance,
                 Current_Bank_TransactionLimit,
                 Current_Bank_PerTransactionLimit,
                 IsOtherBank):

        print("\nWithdraw Money :")

        if CurrentWithDrawMoney <= 0:
            print("Please Enter Valid Amount")
            return False

        if CurrentWithDrawMoney > self.Balance:
            print("Insufficient Account Balance")
            return False

        if self.TransactionCount >= Current_Bank_TransactionLimit:
            print("Transaction Limit Exceeded")
            return False

        if self.WithDrawMoney + CurrentWithDrawMoney > Current_Bank_OneDayLimit:
            print("One Day Withdrawal Limit Exceeded")
            return False

        if CurrentWithDrawMoney > Current_Bank_PerTransactionLimit:
            print("Per Transaction Limit Exceeded")
            return False

        if CurrentWithDrawMoney > Current_Bank_ATMBalance:
            print("ATM Balance Not Available")
            return False

        # 5% charge for using another bank ATM

        Charge = 0

        if IsOtherBank:
            Charge = CurrentWithDrawMoney * 0.05

        TotalAmount = CurrentWithDrawMoney + Charge

        if TotalAmount > self.Balance:
            print("Insufficient Balance Including 5% ATM Charges")
            return False

        # Update account details

        self.Balance -= TotalAmount
        self.TransactionCount += 1
        self.WithDrawMoney += CurrentWithDrawMoney

        print("\nSuccessfully Withdraw Money:", CurrentWithDrawMoney)

        if IsOtherBank:
            print("Other Bank ATM Charges (5%):", Charge)

        print("Total Amount Deducted:", TotalAmount)

        return True

    # Deposit Money

    def Deposit(self, DepositMoney, CurrentBank):

        print("\nDeposit Money :")

        if DepositMoney <= 0:
            print("Please Enter Valid Amount")
            return False

        if self.TransactionCount >= CurrentBank.NumberOfTransactionLimit:
            print("Transaction Limit Exceeded")
            return False

        # Update account and ATM balance

        self.Balance += DepositMoney
        CurrentBank.ATMBalance += DepositMoney
        self.TransactionCount += 1

        print("Successfully Deposited:", DepositMoney)

        return True


# ----------------------------------
# Bank Users
# ----------------------------------

All_User = []

All_User.append(User("Mohit", "SBI", 150000, 0, 1001, 123, 0))
All_User.append(User("Keval", "BOI", 50000, 0, 1002, 456, 0))
All_User.append(User("Raj", "ICICI", 240000, 0, 1003, 789, 0))
All_User.append(User("Meet", "BOI", 10000, 0, 1004, 123, 0))
All_User.append(User("Rahul", "SBI", 6000, 0, 1005, 456, 0))


# ----------------------------------
# All Banks
# ----------------------------------

All_Banks = []

All_Banks.append(Banks("SBI", 5000000, 5, 100000, 40000))
All_Banks.append(Banks("BOI", 3000000, 4, 80000, 30000))
All_Banks.append(Banks("ICICI", 2000000, 3, 70000, 20000))


# ----------------------------------
# Main ATM Program
# ----------------------------------

while True:

    print("\n================================")
    print("          ATM MACHINE")
    print("================================")
    print("1. Admin Login")
    print("2. User Login")
    print("3. Exit")
    print("================================")

    try:
        user = int(input("Enter Your Choice : "))
    except ValueError:
        print("Please Enter Valid Choice")
        continue

    # ----------------------------------
    # Admin Option
    # ----------------------------------

    if user == 1:

        print("\nAdmin Menu")
        print("\n 1 For Show All User : ")
        print("\n 2 For Show All Banks : ")
        print("\n 3 For Add User : ")
        print("\n 4 For Remove User : ")
        print("\n 5 For Add Banks : ")
        print("\n 6 For Remove Banks : ")
        Admin_Choice=int(input())

        if Admin_Choice==1:
            for i in range(0,len(All_User)):
                print(All_User[i].User_Detail())

        elif Admin_Choice==2:
            for i in range(0,len(All_Banks)):
                print(All_Banks[i].Bank_Detail())

        elif Admin_Choice==3:
            print("**********************")
            New_UserName=input("Enter New User Name : ")
            New_UserBankName=input("Enter New User Bank Name : ")
            New_UserBankBalance=int(input("Enter New User Bank  Balance : "))
            New_UserCardNumber=int(input("Enter New User Card Number : "))
            New_UserPINNumber=int(input("Enter New User PIN Number : "))
            All_User.append(User(New_UserName, New_UserBankName, New_UserBankBalance, 0, New_UserCardNumber, New_UserPINNumber, 0))
            print("**********************")

        elif Admin_Choice==4:
            print("Remove User : ")

            for i in range(0,len(All_User)):
                print(f"User Name : {All_User[i].UserName} index : {i}")


            print("Enter Index Number To Delete User : ")
            User_delete_Index_Number=int(input())

            All_User.pop(User_delete_Index_Number)

            
             
        elif Admin_Choice==5:
            print("*****************************")
            New_BankName=input("Enter New Bank Name : ")
            New_ATMBalance=int(input("Enter New ATM Balance : "))
            New_NumberOfTransactionLimit=int(input("Enter Transaction Limit : "))
            New_OneDayLimit=int(input("Enter One Day Limit : "))
            New_PerTransactionLimit=int(input("Enter Per Transaction Limit : "))
            All_Banks.append(Banks(New_BankName,New_ATMBalance, New_NumberOfTransactionLimit, New_OneDayLimit, New_PerTransactionLimit))
            print("*****************************")

        elif Admin_Choice==6:
             print("Remove Banks : ")



    # ----------------------------------
    # User Option
    # ----------------------------------

    elif user == 2:

        print("\nUser Login")

        Current_Card_Number = input("Enter Card Number : ")
        Current_PIN = input("Enter PIN : ")

        Current_User_Index = -1

        # Validate Card Number and PIN

        for i in range(len(All_User)):

            if (str(All_User[i].CardNumber) == Current_Card_Number
                    and str(All_User[i].PIN) == Current_PIN):

                Current_User_Index = i
                break

        if Current_User_Index == -1:

            print("\nInvalid Card Number or PIN")
            continue

        Current_User = All_User[Current_User_Index]

        print("\nWelcome", Current_User.UserName)

        # ----------------------------------
        # User Menu
        # ----------------------------------

        while True:

            print("\n--------------------------------")
            print("1. Withdraw Money")
            print("2. Deposit Money")
            print("3. Show Account Details")
            print("4. Exit")
            print("--------------------------------")

            try:
                Current_Choice = int(input("Enter Your Choice : "))
            except ValueError:
                print("Please Enter Valid Choice")
                continue

            # ----------------------------------
            # Withdraw Money
            # ----------------------------------

            if Current_Choice == 1:

                try:
                    User_Amount_To_WithDraw = int(
                        input("Enter Amount To Withdraw : ")
                    )
                except ValueError:
                    print("Please Enter Valid Amount")
                    continue

                print("\nAvailable ATM Banks")

                for bank in All_Banks:
                    print(bank.BankName)

                User_Enter_ATM_Name = input(
                    "\nEnter ATM Bank Name : "
                ).strip().upper()

                Current_Index_Banks = -1

                # Find Selected ATM Bank

                for i in range(len(All_Banks)):

                    if All_Banks[i].BankName.upper() == User_Enter_ATM_Name:
                        Current_Index_Banks = i
                        break

                if Current_Index_Banks == -1:

                    print("ATM Bank Not Found")
                    continue

                Current_Bank = All_Banks[Current_Index_Banks]

                # Check Other Bank ATM

                IsOtherBank = (
                    Current_User.BankName.upper()
                    != Current_Bank.BankName.upper()
                )

                # Call Withdraw Method

                Result = Current_User.WithDraw(
                    User_Amount_To_WithDraw,
                    Current_Bank.OneDayLimit,
                    Current_Bank.ATMBalance,
                    Current_Bank.NumberOfTransactionLimit,
                    Current_Bank.PerTransactionLimit,
                    IsOtherBank
                )

                # Update ATM Balance

                if Result:

                    Current_Bank.ATMBalance -= User_Amount_To_WithDraw

            # ----------------------------------
            # Deposit Money
            # ----------------------------------

            elif Current_Choice == 2:

                try:
                    DepositMoney = int(
                        input("Enter Deposit Amount : ")
                    )
                except ValueError:
                    print("Please Enter Valid Amount")
                    continue

                # Find User's Own Bank ATM

                Current_Bank = None

                for bank in All_Banks:

                    if bank.BankName == Current_User.BankName:
                        Current_Bank = bank
                        break

                if Current_Bank is None:

                    print("User Bank Not Found")
                    continue

                # Call Deposit Method

                Current_User.Deposit(DepositMoney, Current_Bank)

            # ----------------------------------
            # Account Details
            # ----------------------------------

            elif Current_Choice == 3:

                print("\nAccount Details")

                Current_User.User_Detail()

            # ----------------------------------
            # Exit User Menu
            # ----------------------------------

            elif Current_Choice == 4:

                print("Thank You For Using ATM")
                break

            else:

                print("Invalid Choice")

    # ----------------------------------
    # Exit Program
    # ----------------------------------

    elif user == 3:

        print("\nThank You For Using ATM Machine")
        break

    else:

        print("Invalid Choice")


    
```

## 13) Tic Tac Game

```python
Create a Tic Tac Toe game in python.
1. You should be able to play with two players.
2. You should be able to play against computer.
After you complete simple game implement it in a way that computer never loses.
```

```python
matrix=[
    [1,2,3],
    [4,5,6],
    [7,8,9]
]


def show_matrix():
    print("************************")
    for i in range(0,3,1):
        for j in range(0,3,1):
            print(matrix[i][j],end=" ")
        print()
    print("************************")

def change_matrix(index,sign):
     global matrix
     for i in range(0,3,1):
        for j in range(0,3,1):
            if(index==matrix[i][j]):
                matrix[i][j]=sign


# Row Check Matrix
def check_1(sign):
    for i in range(0,3,1):
        temp=0
        for j in range(0,3,1):
            if matrix[i][j] == sign:
                temp+=1
            
        if temp==3:
            return 1

# Column Check Martix
def check_2(sign):
    for i in range(0,3,1):
        temp=0
        for j in range(0,3,1):
            if matrix[j][i] == sign:
                temp+=1
            
        if temp==3:
            return 1

# Diagonal Element Check Matrix
def check_3(sign):
    temp=0
    for i in range(0,3,1):
        for j in range(0,3,1):
            if matrix[i][j] == sign and i==j:
                temp+=1
            
        if temp==3:
            return 1


# Anti Diagonal Element Check Matrix
def check_3(sign):
    temp=0
    for i in range(0,3,1):
        for j in range(0,3,1):
            if matrix[i][j] == sign and i+j==2:
                temp+=1
            
        if temp==3:
            return 1

count=0
while 1:
    print("-------------------")
    print(f"Player {"1" if count%2==0 else "2" } Sign {"O" if count%2==0 else "X" } ")
    print("Enter Index Number :")
    print("-------------------")
    print(f"Enter Index Number : ")
    show_matrix()
    user_index=int(input())
    sign="O" if count%2==0 else "X"
    change_matrix(user_index,sign)
    show_matrix()
    if check_1(sign) == 1:
        print(f"Winner : Player {"1" if count%2==0 else "2" }")
        break
    if check_2(sign) == 1:
        print(f"Winner : Player {"1" if count%2==0 else "2" }")
        break
    if check_3(sign) == 1:
        print(f"Winner : Player {"1" if count%2==0 else "2" }")
        break
    count+=1
```

## 15)  LCM - GCM

```python
Find LCM and GCM of a number
```

```python

a = int(input("Enter Value 1 : "))
b = int(input("Enter Value 2 : "))

big = 0
small = 0


if a > b:
    big = a
else:
    big = b

# LCM
while 1:
    if big % a == 0 and big % b == 0:
        print(f"LCM : {big}")
        break
    else:
        big += 1

if a > b:
    small = b
else:
    small = a


# GCD
for i in range(1, small + 1):
    if a % i == 0 and b % i == 0:
        gcd = i

print(f"GCD : {gcd}")
```

## 16) Rearrange Array

```python
list1=[1,3,4,2,0]
ans=[0]*len(list1)

print(list1)

for i in range(0,len(list1),1):
    ans[list1[i]]=i

print(ans)
```

## 23) Pattern 7

```python
       1
     2 3 2
   3 4 5 4 3
 4 5 6 7 6 5 4
5 6 7 8 9 8 7 6 5
```

```python
num=int(input("Enter Number : "))

for i in range(1,num+1,1):

    for s in range(0,(num-i),1):
        print(" ",end=" ")
    
    count=i
    for j in range(1,i*2,1):
        if j<=i:
            print(count,end=" ")
            count+=1
        else:
            count-=1
            print(count,end=" ")

        if i==j:
            count-=1
    print()
```

## 25) Pattern 9

      1
     1 1
    1 2 1
   1 3 3 1
  1 4 6 4 1
 1 5 10 10 5 1

```python
num = int(input("Enter Number of Rows: "))

lst = []

for i in range(num):

    row = []

    for j in range(i + 1):

        if j == 0 or j == i:
            row.append(1)

        else:
            value = lst[i-1][j-1] + lst[i-1][j]
            row.append(value)

    lst.append(row)


print(lst)
print(len(lst[0]))
print(len(lst[1]))
print(len(lst[2]))
print(len(lst[3]))

for i in range(num):
    print(" " * (num-i), end="")

    for j in range(i+1):
        print(lst[i][j], end=" ")
    print()
```

## 27) Pattern 11

```python
1 2 3 4
12 13 14 5
11 16 15 6
10 9 8 7
```

```python

n = int(input("Enter Number: "))

# List Compreshions 
arr = [[0 for j in range(n)] for i in range(n)]

top = 0
bottom = n - 1
left = 0
right = n - 1

num = 1

while num <= n * n:

    # 1. Left to Right
    for j in range(left, right + 1):
        arr[top][j] = num
        num += 1

    top += 1

    # 2. Top to Bottom
    for i in range(top, bottom + 1):
        arr[i][right] = num
        num += 1

    right -= 1

    # 3. Right to Left
    for j in range(right, left - 1, -1):
        arr[bottom][j] = num
        num += 1

    bottom -= 1

    # 4. Bottom to Top
    for i in range(bottom, top - 1, -1):
        arr[i][left] = num
        num += 1

    left += 1


for i in range(n):
    for j in range(n):
        print(arr[i][j], end=" ")

    print()
```

## 28) Pattern 12 (Diamond Pattern)

```python
num=int(input("Enter Number : "))

for i in range(1,num+1,1):

    for s in range(0,num-i,1):
        print(" ",end="")
    
    for j in range(1,(i+1),1):
        print("*",end=" ")
        
    print()

for i in range(num-1,0,-1):

    for s in range(0,num-i,1):
        print(" ",end="")
    
    for j in range(1,(i+1),1):
        print("*",end=" ")
    print()

```
