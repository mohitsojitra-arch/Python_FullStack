#### create function based script here

```python
def Marks_CalCulation():
    stu_name = input("Enter Student Name: ")

    marks = []

    for i in range(0, 5):
        marks.append(int(input(f"Enter Subject {i+1} Marks: ")))

    for i in marks:
        if i < 23:
            print("Fail!")
            break
    else:
        percentage = (sum(marks) / (len(marks) * 70)) * 100
        print(f"Student Name: {stu_name}")
        print(f"Percentage: {round(percentage, 2)}%")


Marks_CalCulation()
```

#### create class based script here

```python
class Result:
    def Marks_CalCulation(self):
        stu_name = input("Enter Student Name: ")
        marks = []
    
        for i in range(0, 5):
            marks.append(int(input(f"Enter Subject {i+1} Marks: ")))
    
        for i in marks:
            if i < 23:
                print("Fail!")
                break
        else:
            percentage = (sum(marks) / (len(marks) * 70)) * 100
            print(f"Student Name: {stu_name}")
            print(f"Percentage: {round(percentage, 2)}%")


student1=Result()
student1.Marks_CalCulation()
```

## Implement stack

```python
stack=list()
current_count=1

def push():
   global current_count
   if current_count<=count: 
    num=int(input(f"Enter Stack Element :  {current_count}/{count}"))
    current_count+=1
    stack.append(num)
   else:
     print("Stack Is Overflow : ")

def pop():
    global current_count
    if current_count==1:
        print("Stack Is Underflow (Empty) : ")
    else:
        stack.pop()
        current_count-=1
        print("POP Element : ")

def show():
    print(stack)

def peek():
    if current_count == 1:
        print("Stack Is Underflow (Empty) : ")
    else:
        print(stack[len(stack)-1])

        

count=int(input("Enter Stack Size : "))

while 1:
    print("1 For PUSH() \n 2 For POP() \n 3 For SHOW() \n 4 For PEEK() \n 5 For Exit")
    user=int(input("Enter Option : "))

    if user == 1:
     push()
    elif user == 2:
     pop()
    elif user == 3:
     show()
    elif user == 4:
      peek()
    elif user == 5:
        break
    else:
        print("Enter Valid Option : ")
```

## Implement queue

```python
list1=list()
size=int(input("Enter Queue Size : "))
current_size=1

def enqueue():
    global current_size
    global size
    if  current_size<=size:
        num=int(input(f"Enter Element : {current_size}/{size}"))
        list1.append(num)
        current_size+=1
    else:
        print("Queue Overflow : ")


def dequeue():
    global current_size
    global size
    if current_size==1:
        print("Queue in Underflow : ")
    else:
        ele=list1.pop(0)
        current_size-=1
        print(f"Dequeue Element : {ele}")
        

def show():
    print(list1)


def peek():
    if current_size==1:
        print("Queue Is Underflow")
    else:
        print(f"Peek Element :{list1[0]}")



while 1:
    print("Choose Option \n 1 For Enqueue \n 2 For Dequeue \n 3 For Show \n 4 For Peek \n 5 For Exit")
    num=int(input("Enter Number : "))

    if num == 1:
        enqueue()
    elif num==2:
        dequeue()
    elif num==3:
        show()
    elif num==4:
        peek()
    elif num==5:
        break
    
```

## Implement dqueue

```python
list1=list()

size=int(input("Enter Size Of Circular Queue : "))
current_size=1


def FrontEnqueue():
    global current_size
    if current_size<=size:
        num=int(input(f"Enter Element (Front-Add) : {current_size}/{size}"))
        list1.insert(0,num)
        current_size+=1
    else:
        print("Circular Queue Is Overflow : ")


def RearEnqueue():
    global current_size
    if current_size<=size:
        num=int(input(f"Enter Element (Rear-Add) : {current_size}/{size}"))
        list1.append(num)
        current_size+=1
    else:
        print("Circular Queue Is Overflow : ")


def FrontDequeue():
    global current_size
    if current_size==1:
        print("Circular Queue Is Underflow : ")
    else:
        n=list1.pop(0)
        print(f"Front Dequeue : {n}")
        current_size-=1


def RearDequeue():
    global current_size
    if current_size==1:
        print("Circular Queue Is Underflow : ")
    else:
        n=list1.pop()
        print(f"Rear Dequeu : {n}")
        current_size-=1


def show():
    print(list1)

while 1:
    print("Choose Number \n 1 For FrontEnqueue \n 2 For RearEnqueue \n 3 For FrontDequeue \n 4 For RearDequeue \n 5 For Show \n 6 For Exit")
    user=int(input("Enter Option : "))

    if user==1:
        FrontEnqueue()
    elif user==2:
        RearEnqueue()
    elif user==3:
        FrontDequeue()
    elif user==4:
        RearDequeue()
    elif user==5:
        show()
    elif user==6:
        break 
    else:
        print("Enter Valid Option : ")
        
        
```

## Sorting algorithms

#### Selection Sort

```python
def selection_sort(items):
    n = len(items)
    
    for i in range(n - 1):
        min_idx = i
        for j in range(i + 1, n):
            if items[j] < items[min_idx]:
                min_idx = j
                
        if min_idx != i:
            items[i], items[min_idx] = items[min_idx], items[i]
            
    return items

my_list = [64, 34, 25, 12, 22, 11, 90]
sorted_list = selection_sort(my_list)
print("Sorted List:", sorted_list)
```

#### Bubble Sort

```python
def bubble_sort(arr):
    n = len(arr)
    
    for i in range(n):
        swapped = False
        

        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
                
        if not swapped:
            break

numbers = [64, 34, 25, 12, 22, 11, 90]
print("Original list:", numbers)

bubble_sort(numbers)
print("Sorted list:  ", numbers)
```

#### Insertion Sort

```python
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i] 
        j = i - 1    
      
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
            
       
        arr[j + 1] = key


data = [12, 11, 13, 5, -1]
insertion_sort(data)
print("Sorted array:", data)
```

#### Merge Sort

```python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2
    left_half = merge_sort(arr[:mid])
    right_half = merge_sort(arr[mid:])

    return merge(left_half, right_half)

def merge(left, right):
    sorted_list = []
    i = j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            sorted_list.append(left[i])
            i += 1
        else:
            sorted_list.append(right[j])
            j += 1

    sorted_list.extend(left[i:])
    sorted_list.extend(right[j:])
    
    return sorted_list

numbers = [38, 27, 43, 3, 9, 82, 10]
print(merge_sort(numbers))  
```

#### Quick Sort

```python
def partition(arr, low, high):
    pivot = arr[high]
    i = low - 1
    for j in range(low, high):
        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]
    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return i + 1

def quick_sort_inplace(arr, low=0, high=None):
    if high is None:
        high = len(arr) - 1
    if low < high:
        pi = partition(arr, low, high)
        quick_sort_inplace(arr, low, pi - 1)
        quick_sort_inplace(arr, pi + 1, high)

numbers = [10, 7, 8, 9, 1, 5]
quick_sort_inplace(numbers)
print(numbers) 
```

## Searching Algorithms

#### Linear Search

```python
def Linear_Search(arr,target):

    for i in range(0,len(arr),1):
        if arr[i]==target:
            return i


    return -1

arr=[10,20,30,40,50,60,70]

print(Linear_Search(arr,50))
```

##### Binary Search

```python
def binary_search(arr, target):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2
        

        if arr[mid] == target:
            return mid
       
        elif arr[mid] < target:
            low = mid + 1
      
        else:
            high = mid - 1
            
   
    return -1


mylist = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
print(binary_search(mylist, 23))  
```
